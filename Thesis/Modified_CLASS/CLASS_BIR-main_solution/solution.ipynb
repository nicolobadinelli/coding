```python
import numpy as np
import matplotlib.pyplot as plt

# Load data
d = np.loadtxt("output/scalaron_cl.dat")

# Extract ell
ell = d[:,0]

# Conversion factors
rad_to_deg = 180/np.pi

# Extract spectra
claa = d[:,7] * rad_to_deg**2     # C_ell^{alpha alpha}
claT = d[:,10] * rad_to_deg       # C_ell^{alpha T}
claE = d[:,13] * rad_to_deg       # C_ell^{alpha E}

# Prefactor
pref = ell * (ell + 1) / (2 * np.pi)

# Apply transformations
claa_plot = pref * claa
claT_plot = pref * np.abs(claT)   # absolute value
claE_plot = pref * np.abs(claE)   # absolute value

# =========================
# 1. C_ell^{alpha alpha}
# =========================
plt.figure(figsize=(8,6))
plt.loglog(ell, claa_plot)

plt.xlabel(r"$\ell$")
plt.ylabel(r"$\ell(\ell+1)C_\ell^{\alpha\alpha}/(2\pi)$")
plt.title(r"$C_\ell^{\alpha\alpha}$")
plt.xlabel(r"$\ell$")
plt.ylabel(r"$\ell(\ell+1)C_\ell^{\alpha\alpha}/2\pi$ $\text{deg}^2$")
plt.ylim(1e-48, 1e0)
plt.xlim(2, 2500)
plt.grid(which='both', linestyle='--', linewidth=0.5, alpha=1)
plt.show()


# =========================
# 2. |C_ell^{alpha T}|
# =========================
plt.figure(figsize=(8,6))
plt.loglog(ell, claT_plot)

plt.xlabel(r"$\ell$")
plt.ylabel(r"$\ell(\ell+1)|C_\ell^{\alpha T}|/(2\pi)$")
plt.title(r"$|C_\ell^{\alpha T}|$")

plt.xlim(2, 2500)
plt.ylim(1e-25, 1e-1)
plt.xlabel(r"$\ell$")
plt.ylabel(r"$\ell(\ell+1)|C_\ell^{\alpha T}|/2\pi$ $\text{deg}\cdot\mu\text{K}$")
plt.grid(which='both', linestyle='--', linewidth=0.5, alpha=1)
plt.legend(fontsize=10, frameon=True)
plt.show()


# =========================
# 3. |C_ell^{alpha E}|
# =========================
plt.figure(figsize=(8,6))
plt.loglog(ell, claE_plot)

plt.xlabel(r"$\ell$")
plt.ylabel(r"$\ell(\ell+1)|C_\ell^{\alpha E}|/(2\pi)$")
plt.title(r"$|C_\ell^{\alpha E}|$")
plt.xlim(2, 2500)
plt.ylim(1e-25, 1e-1)
plt.xlabel(r"$\ell$")
plt.ylabel(r"$\ell(\ell+1)|C_\ell^{\alpha E}|/2\pi$ $\text{deg}\cdot\mu\text{K}$")
plt.grid(which='both', linestyle='--', linewidth=0.5, alpha=1)
plt.legend(fontsize=10, frameon=True)
plt.show()
```
